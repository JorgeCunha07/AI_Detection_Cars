## 1. Introdução e Visão Geral

O projeto do **ChatBot do Código da Estrada Português** é uma aplicação interativa desenvolvida para facilitar a aprendizagem e consulta do Código da Estrada de Portugal. A solução integra um modelo de linguagem treinado especificamente para o domínio rodoviário português, permitindo aos utilizadores interagirem através de linguagem natural para obter informações, consultar artigos específicos do código, e testar os seus conhecimentos através de um modo quiz.

### Objectivos do Projecto

1. **Facilitar o acesso à informação**: Permitir consultas em linguagem natural sobre regras de trânsito
2. **Assistir na aprendizagem**: Oferecer um modo quiz com perguntas e respostas sobre o Código da Estrada
3. **Consulta rápida**: Acesso direto a artigos específicos do Código da Estrada
4. **Experiência natural**: Reconhecer variações na forma como os utilizadores expressam respostas

### Problema Abordado

O estudo e memorização do Código da Estrada é tradicionalmente um processo baseado em leitura passiva e memorização. Este chatbot transforma essa experiência numa interação dinâmica, permitindo que os utilizadores testem os seus conhecimentos e obtenham respostas imediatas às suas dúvidas, num formato conversacional e acessível.

## 2. Arquitectura do Sistema

O sistema está estruturado em vários componentes interligados, seguindo uma arquitectura modular que permite a evolução e melhoria independente de cada componente.

### Diagrama de Arquitectura


### Fluxo de Dados

1. O utilizador introduz uma consulta ou resposta a uma pergunta
2. O sistema processa o input através de um dos seguintes caminhos:
   - **Modo Conversacional**: Encaminha a consulta para o modelo linguístico que gera uma resposta
   - **Modo Quiz**: Compara a resposta com as respostas correctas usando o sistema de similaridade
   - **Modo Consulta**: Busca diretamente em artigos do Código da Estrada
3. O sistema apresenta o resultado ao utilizador de forma formatada

```python
    input_ids = self.tokenizer.encode(prompt, return_tensors="pt").to(self.device)
    with torch.no_grad():
        output = self.model.generate(

            max_length=len(input_ids[0]) + max_length,

        )
```

## 7. Comparação de Abordagens de Similaridade

O projeto implementa e compara três métodos diferentes para verificar a similaridade entre a resposta do utilizador e as respostas corretas:

### Método 1: Similaridade Simples (Baseline)

Presente no `chat_interativo.py`, este método básico utiliza interseção de palavras para determinar similaridade.

```python
def similar(self, texto1, texto2, threshold=0.7):
    palavras1 = set(texto1.split())
    palavras2 = set(texto2.split())
    intersection = palavras1.intersection(palavras2)
    return len(intersection) / max(len(palavras1), len(palavras2)) >= threshold
```

**Limitações:** Não reconhece sinónimos, variações de formato ou estruturas de frase diferentes.

### Método 2: Sistema Baseado em Regras (Avançado)

Implementado em `chat_interativo_fixed.py`, este método utiliza um conjunto de regras e um dicionário de sinónimos para identificar respostas similares.

**Características:**
- Dicionário de sinónimos para palavras-chave
- Normalização de formatos (velocidade, distância)
- Regras específicas para diferentes tipos de perguntas
- Ponderação especial para números e unidades

**Vantagens:** Bom desempenho em casos específicos onde as regras foram otimizadas manualmente.

**Limitações:** Requer manutenção manual do dicionário de sinónimos e regras, o que se torna impraticável à medida que o sistema cresce.

### Método 3: Sistema Baseado em Transformers (Estado da Arte)

Implementado em `transformer_similarity.py` e utilizado em `chat_interativo_transformer.py`, este método utiliza modelos transformer pré-treinados para calcular a similaridade semântica entre frases.

```python

    return similarity_score >= threshold
```

**Vantagens:**
- Compreensão semântica superior
- Reconhece variações naturais de expressão
- Não requer manutenção manual de sinónimos
- Generaliza melhor para novos casos

**Resultados da Comparação:**

Usando o script `comparar_sistemas.py`, foram testadas diversas variações de respostas para diferentes categorias de perguntas (velocidade, distância, ultrapassagem, estacionamento). Os resultados mostram:

- **Sistema Transformer:** ~92% de precisão geral
- **Sistema Baseado em Regras:** ~78% de precisão geral

A maior diferença foi observada na capacidade de reconhecer variações naturais de expressão, onde o sistema transformer reconheceu corretamente frases que expressavam o mesmo conceito de maneiras muito diferentes.

## 8. Análise de Desempenho

### Métricas de Desempenho

Os testes realizados com `comparar_sistemas.py` revelaram as seguintes métricas:

| Sistema             | Taxa Acerto Respostas Corretas | Taxa Rejeição Respostas Incorretas | Tempo Médio de Processamento |
|---------------------|--------------------------------|-------------------------------------|------------------------------|
| Baseado em Regras   | 78%                            | 89%                                 | 0.002s                       |
| Transformer         | 92%                            | 96%                                 | 0.15s                        |

### Análise de Recursos

O sistema transformer requer mais recursos computacionais, mas oferece resultados superiores:

| Sistema             | Uso de Memória | Suporte a GPU    | Tempo de Inicialização |
|---------------------|----------------|------------------|------------------------|
| Baseado em Regras   | ~100MB         | Não necessário   | <1s                    |
| Transformer         | ~500-800MB     | Recomendado      | 3-5s                   |

### Escalabilidade

O sistema transformer é mais escalável a longo prazo, pois não requer atualizações manuais de regras e dicionários quando novos conteúdos são adicionados. O sistema baseado em regras precisaria de constantes atualizações para manter a sua eficácia.
