Validar a qualidade de **texto gerado automaticamente por modelos de linguagem**, como o do teu chatbot, pode ser desafiante, mas existem **várias abordagens objetivas e semiautomáticas** que te podem ajudar. Abaixo apresento uma combinação de **métricas automáticas, avaliação humana e critérios linguísticos** para validação do output:

---

### ✅ **1. Métricas automáticas (quantitativas)**

Estas comparam o texto gerado com uma ou mais respostas de referência:

#### a) **BLEU (Bilingual Evaluation Understudy)**
- Mede a sobreposição de n-gramas entre o output e a resposta correta.
- Muito usada em tradução automática.
- Limitação: sensível a sinónimos e variações de linguagem.

```python
from nltk.translate.bleu_score import sentence_bleu
score = sentence_bleu([referencia.split()], gerado.split())
```

#### b) **ROUGE (Recall-Oriented Understudy for Gisting Evaluation)**
- Mede a sobreposição de n-gramas, especialmente para sumarização de texto.
- ROUGE-1 (unigramas), ROUGE-2 (bigramas), ROUGE-L (longest common subsequence).

```python
from rouge import Rouge
rouge = Rouge()
scores = rouge.get_scores(gerado, referencia)
```

#### c) **BERTScore**
- Mede a similaridade semântica entre dois textos usando embeddings de BERT.
- Muito útil em textos curtos ou com sinónimos.

```python
from bert_score import score
P, R, F1 = score([gerado], [referencia], lang="pt", model_type="neuralmind/bert-base-portuguese-cased")
```

### 🧠 **2. Avaliação semântica com embeddings**

Caso não tenhas uma referência direta, podes usar modelos como **SBERT** para calcular a **similaridade de cosseno** entre o texto gerado e os documentos/artigos do Código da Estrada.

```python
from sentence_transformers import SentenceTransformer, util
model = SentenceTransformer("paraphrase-multilingual-MiniLM-L12-v2")

emb1 = model.encode(gerado, convert_to_tensor=True)
emb2 = model.encode(referencia, convert_to_tensor=True)
similaridade = util.cos_sim(emb1, emb2)
```

### 🧑‍⚖️ **3. Avaliação humana com critérios objetivos**

Cria uma grelha de avaliação com 4-5 critérios e pontua de 1 a 5 (ou 0 a 10):

| Critério               | Descrição                                               | Escala |
|------------------------|---------------------------------------------------------|--------|
| Correcção              | A resposta está factualmente correta?                   | 0-5    |
| Clareza                | É fácil de ler e entender?                              | 0-5    |
| Coerência              | O texto faz sentido e segue um raciocínio lógico?       | 0-5    |
| Utilidade/Relevância   | A resposta responde à pergunta de forma útil?           | 0-5    |
| Fluência linguística   | A gramática e ortografia estão corretas?                | 0-5    |
