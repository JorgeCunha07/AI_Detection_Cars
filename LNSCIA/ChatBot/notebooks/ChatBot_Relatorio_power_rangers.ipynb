# 🤖 Estrutura Modular do ChatBot do Código da Estrada

Dividimos o chatbot em **3 módulos especializados**, porque tentar fazer tudo com um modelo só é pedir um acidente de viação computacional.

---

## 🧠 Por que dividir o bot?

### 1. Especialização melhora a performance
Cada parte tem foco numa função:
- Conversa aberta
- Validação de respostas
- Pesquisa legal

### 2. Eficiência
Evita carregar modelos pesados para tarefas simples.

---

## 🧩 Os 3 Módulos Explicados

---

### 🗨️ 1. Conversa Natural – `chat_interativo_transformer.py`

**Função:**  
Conversar de forma fluida sobre o Código da Estrada.

**Tecnologias:**
- **Modelo:** `pierreguillou/gpt2-small-portuguese`
- **Técnica:** *Fine-tuning* com `dialogos.json`
- **Tipo:** Causal Language Modeling (CLM)

**Uso:**  
Chat tipo ChatGPT. Ideal para perguntas abertas e explicações.

---

### ✅ 2. Validação de Respostas – `model_interface.py`

**Função:**  
Faz perguntas tipo teste de código e valida as tuas respostas.

**Tecnologias:**
- Carrega perguntas de `questoes.json`
- Verifica respostas por lógica direta
- Pode usar embeddings no futuro para feedback automático

**Uso:**  
Testes práticos, quizzes, e trauma emocional pré-IMT.

---

### 📚 3. Verificação de Artigos – `similar_semantic.py`

**Função:**  
Dada uma pergunta, encontra o artigo mais relevante do Código da Estrada.

**Tecnologias:**
- **Modelo:** `sentence-transformers/paraphrase-multilingual-MiniLM-L12-v2`
- **Técnica:** Sentence Embedding + Cosine Similarity
- **Dados:** `Codigo_Estrada.json`

**Uso:**  
Pesquisa semântica legal. Parece magia negra, mas é só álgebra linear sexy.

---

## 🧃 Resumo da Estrutura

| Módulo        | Função                       | Tecnologia                              | Serve para                         |
|---------------|------------------------------|------------------------------------------|-------------------------------------|
| Conversa      | Conversa aberta              | GPT2 CLM, Fine-tuned                     | Perguntar e explicar                |
| Validação     | Corrigir perguntas/respostas | Scripts, lógica simples, JSON            | Simular testes                      |
| Verificação   | Buscar artigos relevantes    | SBERT + Cosine Similarity                | Pesquisa legal                      |
