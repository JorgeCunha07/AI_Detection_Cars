```python
# Jupyter Notebook - Convert BDD100K to YOLOv5 / YOLOv4 / YOLOX

# 📁 Etapa 1: Defina o caminho do dataset BDD100K
source_dir = "/Users/jorgecunha/PycharmProjects/AI_Detection_Cars/AAU2/Test/bdd100k"

# 🧠 Etapa 2: Importar e Analisar com FiftyOne
import fiftyone as fo
import fiftyone.zoo as foz

# Carrega o split de validação (ou use "train" para treino)
dataset = foz.load_zoo_dataset(
    "bdd100k",
    split="validation",  # ou "train"
    source_dir=source_dir,
    copy_files=False,
)

# 🔍 Lançar o app do FiftyOne para inspecionar as imagens (opcional)
session = fo.launch_app(dataset)

# 🛠️ Etapa 3: Converter para formato YOLOv5 (ou outro)
# Você pode trocar para:
# fo.types.YOLOv5Dataset, fo.types.YOLOv4Dataset, fo.types.VOCDetectionDataset, fo.types.COCODetectionDataset

export_dir = "/Users/jorgecunha/PycharmProjects/AI_Detection_Cars/AAU2/Test/bdd100k_yolo_converted"
dataset_type = fo.types.YOLOv5Dataset

# Exportação
dataset.export(
    export_dir=export_dir,
    dataset_type=dataset_type,
    label_field="detections",  # campo padrão de detecção do BDD100K
    export_media="symlink",    # ou "copy" se quiser copiar imagens
)

print(f"✅ Exportação concluída para {export_dir}")
```
